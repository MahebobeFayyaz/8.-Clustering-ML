```python
import pandas as pd

data = {
    "Customer": [
        "A","B","C","D","E","F","G","H"
    ],
    "Annual Income": [
        15, 18, 20, 60, 65, 70, 120, 125
    ],
    "Spending Score": [
        80, 85, 75, 40, 45, 35, 10, 15
    ]
}

small_df = pd.DataFrame(data)

small_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(
    small_df["Annual Income"],
    small_df["Spending Score"],
    s=100
)

for i, txt in enumerate(small_df["Customer"]):
    plt.annotate(
        txt,
        (
            small_df["Annual Income"][i],
            small_df["Spending Score"][i]
        ),
        fontsize=12
    )

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")

plt.title("Customer Data")

plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

X = small_df[
    [
        "Annual Income",
        "Spending Score"
    ]
]


scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
import scipy.cluster.hierarchy as sch

linkage_matrix = sch.linkage(
    X_scaled,
    method="ward"
)
```

```python
plt.figure(figsize=(10,5))

sch.dendrogram(
    linkage_matrix,
    labels=small_df["Customer"].values
)

plt.title("Hierarchical Clustering Dendrogram")

plt.xlabel("Customers")

plt.ylabel("Distance")

plt.show()
```

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

sch.dendrogram(
    linkage_matrix,
    labels=small_df["Customer"].values
)

# Draw horizontal cut line at distance 1.5
plt.axhline(
    y=1.5,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Cut at distance = 1.5"
)

plt.title("Dendrogram with Cut Line")

plt.xlabel("Customers")

plt.ylabel("Distance")

plt.legend()

plt.show()
```
